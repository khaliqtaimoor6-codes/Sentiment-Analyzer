Text Cleaning
1. Sample 10000 rows
2. Remove html tags
3. Remove special characters
4. Converting everything to lowercase
5. Removing stop words (ie and, is, the)
6. Stemming (ie playing as play)

```python
df = df.sample(10000)
```

```python
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df['sentiment'] = df['sentiment'].map({
    'positive': 1,
    'negative': 0
})
```

```python
print(df['sentiment'])
```

```python
import re
clean = re.compile('<.*?>')
re.sub(clean, '', df.iloc[2].review)
```

```python
def clean_html(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)
```

```python
df['review'] = df['review'].apply(clean_html)
```

```python
def convert_lower(text):
    return text.lower()
```

```python
df['review'] = df['review'].apply(convert_lower)
```

```python
def remove_special(text):
    x = ''
    for i in text:
        if i.isalnum():
            x = x + i
        else:
            x = x + ' '
    return x
```

```python
df['review'] = df['review'].apply(remove_special)
```

```python
import nltk
from nltk.corpus import stopwords
```

```python
def remove_stopwords(text):
    x = []
    for i in text.split():
        if i not in stopwords.words('english'):
            x.append(i)
    y = x[:]
    x.clear()
    return y
```

```python
df['review'] = df['review'].apply(remove_stopwords)
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
y = []

def stem_words(text):
    for i in text:
        y.append(ps.stem(i))
    z = y[:]
    y.clear()
    return z
```

```python
df['review'] = df['review'].apply(stem_words)
```

```python
def join_back(list_input):
    return " ".join(list_input)
```

```python
df['review'] = df['review'].apply(join_back)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```

```python
X = cv.fit_transform(df['review']).toarray()
```

```python
X.shape
```

```python
X[0].max()
```

```python
y = df.iloc[:, -1].values
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB

clf1 = GaussianNB()
clf2 = MultinomialNB()
clf3 = BernoulliNB()
```

```python
clf1.fit(X_train, y_train)
clf2.fit(X_train, y_train)
clf3.fit(X_train, y_train)
```

```python
y_pred1 = clf1.predict(X_test)
y_pred2 = clf2.predict(X_test)
y_pred3 = clf3.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print("Gaussian NB: ", accuracy_score(y_test, y_pred1))
print("Multinomial NB: ", accuracy_score(y_test, y_pred2))
print("Bernoulli NB: ", accuracy_score(y_test, y_pred3))
```
